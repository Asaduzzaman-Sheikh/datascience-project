```python
ethnic_group = df.groupby('race/ethnicity')[['average score', 'math score', 'reading score', 'writing score']].mean()
group_A = ethnic_group.loc['group A']
group_B = ethnic_group.loc['group B']
group_C = ethnic_group.loc['group C']
group_D = ethnic_group.loc['group D']
group_E = ethnic_group.loc['group E']

plt.figure(figsize=(16, 9)) # Slightly wider layout gives text more breathing room

X = ["Average Score", "Math Score", "Reading Score", "Writing Score"]
X_axis = np.arange(len(X))

# 1. Shrink width to 0.15 so 5 bars fit perfectly without overlapping
width = 0.15

# 2. Perfect mathematical offsets to center all 5 bars over the tick marks
plt.bar(X_axis - (2 * width), group_A, width, label='Group A', color='#1f77b4')
plt.bar(X_axis - width,       group_B, width, label='Group B', color='#ff7f0e')
plt.bar(X_axis,               group_C, width, label='Group C', color='#2ca02c')
plt.bar(X_axis + width,       group_D, width, label='Group D', color='#d62728')
plt.bar(X_axis + (2 * width), group_E, width, label='Group E', color='#9467bd')

plt.xticks(X_axis, X, fontsize=12)
plt.ylabel("Marks", fontsize=12)
plt.title("Score Breakdown by Subject and Race/Ethnicity", fontweight='bold', fontsize=16)

# 3. Dynamic adjustment to make sure the Y-axis has room for text labels above the highest bar
plt.ylim(0, df[['average score', 'math score', 'reading score', 'writing score']].max().max() + 10)

plt.legend(loc='upper right')

# 4. Clean container text placement with slightly smaller font size to avoid bumping sides
for container in plt.gca().containers:
    plt.gca().bar_label(container, fmt='%.1f', padding=4, fontsize=9, fontweight='semibold')

plt.tight_layout()
plt.show()
```
